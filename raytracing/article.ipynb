```python
import sionna
from sionna.rt import load_scene, Camera
from IPython.display import display

# Load integrated scene
scene = load_scene(sionna.rt.scene.)

# Open interactive 3D preview
scene.preview()

# Render scene from preview viewpoint
scene.render_to_file(
    camera="preview",
    filename="scene.png"
)

# Create new camera with different configuration
my_cam = Camera(
    position=(0, 20, 30),
    look_at=(0, 0, 0)
)

# Render scene with new camera
image = scene.render(camera=my_cam)
display(image)

# Render scene to file in high resolution
scene.render_to_file(
    camera=my_cam,
    filename="scene2.png",
    resolution=(1920, 1080)
)
```
